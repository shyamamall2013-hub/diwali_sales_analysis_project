# Exploratory Data Analysis


# Gender

```python
ax =sns.countplot(data = df, x = "Gender")
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Total Amount vs Gender

sales_gen = df.groupby(["Gender"], as_index = False).agg({"Amount":"sum"})
print(sales_gen)
sns.barplot(data = sales_gen, x = "Gender", y = "Amount")
```

```python
ax = sns.countplot(data = df, x = "Age_Group",hue = "Gender"  )
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Total Amount vs Age Group

sales_age = df.groupby(["Age_Group"], as_index = False).agg({"Amount":"sum"}).sort_values(by = "Amount", ascending = False)
print(sales_age)
sns.barplot(data = sales_age, x = "Age_Group", y = "Amount")
```

# State

```python
# Total number of Orders from top 10 States

sales_state = df.groupby(["State"], as_index = False).agg({"Orders":"sum"}).sort_values(by = "Orders", ascending = False).head(10)
print(sales_state)
sns.barplot(data = sales_state, x = "State", y = "Orders")
sns.set(rc={"figure.figsize":(20,5)})
```

```python
# Total Amount/Sales from top 10 States

gp = df.groupby(["State"], as_index = False).agg({"Amount":"sum"}).sort_values(by = "Amount", ascending = False).head(10)
print(gp)
sns.barplot(data = gp, x = "State", y = "Amount")
sns.set(rc={"figure.figsize":(15,5)})
```

# Marital status

```python
# Total Amount/Sales from top 10 States

ax =sns.countplot(data = df, x = "M_Status_Name")
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_marital_status = df.groupby(["M_Status_Name"], as_index = False).agg({"Amount":"sum"})
print(sales_marital_status)
sns.barplot(data = sales_marital_status, x = "M_Status_Name", y = "Amount")
```

# Occupation

```python
sns.set(rc={"figure.figsize":(20,5)})
order = df["Occupation"].value_counts().sort_values(ascending = False).index
ax =sns.countplot(data = df, x = "Occupation", order=order)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_Occupation = df.groupby(["Occupation"], as_index = False).agg({"Amount":"sum"}).sort_values(by ="Amount", ascending = False)
print(sales_Occupation)
sns.barplot(data = sales_Occupation, x = "Occupation", y = "Amount")
```

# Product Category

```python
sns.set(rc={"figure.figsize":(35,7)})
order = df["Product_Category"].value_counts().sort_values(ascending = False).index
ax =sns.countplot(data = df, x = "Product_Category", order =order)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sns.set(rc={"figure.figsize":(35,7)})
sales_Product_Category = df.groupby(["Product_Category"], as_index = False).agg({"Amount":"sum"}).sort_values(by ="Amount", ascending = False)
print(sales_Product_Category)
sns.barplot(data = sales_Product_Category, x = "Product_Category", y = "Amount")
```

```python
# Top 10 most sold products

sns.set(rc={"figure.figsize":(15,5)})
sales_Product = df.groupby(["Product_Id"], as_index = False).agg({"Amount":"sum"}).sort_values(by ="Amount", ascending = False).head(10)
print(sales_Product)
sns.barplot(data = sales_Product, x = "Product_Id", y = "Amount")
```

```python
# Top 10 most sold products (same thing as above)

fig1, ax1 = plt.subplots(figsize=(7,5))
df.groupby("Product_Id")["Orders"].sum().nlargest(10).sort_values(ascending = False).plot(kind="bar")
```
